```python
# Imports
import pandas as pd
import numpy as np
import joblib
```

```python
#Loading saved model 

model = joblib.load("crop_recommendation_model.pkl")
print("Model loaded successfully.")
```

```python
# Creating multiple test cases with different soil and environmental conditions
# to check whether the trained model can generate crop recommendations.

test_cases = [
    # N, P, K, temperature, humidity, ph, rainfall

    # Normal / balanced conditions
    [90, 42, 43, 26.0, 80.0, 6.5, 200.0],
    [50, 45, 40, 25.0, 80.0, 6.0, 500.0],
    [70, 50, 45, 24.0, 75.0, 6.2, 700.0],
    [100, 60, 50, 27.0, 70.0, 6.5, 900.0],
    [40, 30, 30, 22.0, 65.0, 5.8, 300.0],

    # Low nitrogen
    [5, 40, 40, 25.0, 80.0, 6.0, 500.0],
    [10, 50, 45, 24.0, 75.0, 6.2, 800.0],
    [20, 35, 30, 23.0, 70.0, 5.8, 400.0],

    # High nitrogen
    [120, 50, 45, 26.0, 80.0, 6.5, 600.0],
    [150, 60, 50, 28.0, 75.0, 6.8, 800.0],

    # High phosphorus
    [60, 100, 50, 25.0, 75.0, 6.0, 500.0],
    [80, 130, 60, 27.0, 70.0, 6.5, 700.0],

    # High potassium
    [60, 45, 100, 25.0, 80.0, 6.2, 600.0],
    [80, 60, 150, 28.0, 75.0, 6.5, 800.0],

    # Low rainfall
    [50, 40, 40, 25.0, 60.0, 6.0, 50.0],
    [70, 50, 45, 28.0, 65.0, 6.2, 100.0],

    # High rainfall
    [60, 45, 40, 25.0, 85.0, 6.0, 1200.0],
    [80, 50, 50, 27.0, 90.0, 6.5, 1800.0],

    # Acidic soil
    [60, 40, 40, 25.0, 75.0, 4.5, 500.0],
    [80, 50, 45, 26.0, 80.0, 5.0, 700.0],

    # Near-neutral soil
    [70, 50, 45, 25.0, 75.0, 7.0, 600.0],
    [80, 60, 50, 27.0, 80.0, 7.2, 800.0],

    # Higher pH
    [60, 45, 40, 25.0, 70.0, 8.0, 500.0],
    [90, 50, 50, 28.0, 75.0, 8.5, 700.0]
]

columns = [
    "N",
    "P",
    "K",
    "temperature",
    "humidity",
    "ph",
    "rainfall"
]

test_df = pd.DataFrame(test_cases, columns=columns)

predictions = model.predict(test_df)

test_df["Predicted Crop"] = predictions

print(test_df.to_string(index=False))
```

```python
# Loading the dataset to check on the dataset itself 

df = pd.read_csv("processed_crop_dataset.csv")
```

```python
# Selecting random samples from the original dataset to verify
# whether the saved model correctly predicts their known crop labels.

features = [
    "N",
    "P",
    "K",
    "temperature",
    "humidity",
    "ph",
    "rainfall"
]

random_samples = df.sample(20, random_state=42)

actual = random_samples["label"]
predicted = model.predict(random_samples[features])

comparison = random_samples[features].copy()
comparison["Actual Crop"] = actual.values
comparison["Predicted Crop"] = predicted
comparison["Correct"] = comparison["Actual Crop"] == comparison["Predicted Crop"]

print(comparison.to_string(index=False))
```

```python
random_samples = df.sample(100, random_state=100)

actual = random_samples["label"]
predicted = model.predict(random_samples[features])

accuracy = (actual.values == predicted).mean()

print("Random 100 Sample Accuracy:", accuracy)
```

```python
# Testing the model with extreme values within the observed dataset range

extreme_tests = [
    [0, 5, 5, 1.2, 12, 3.5, 3.3],
    [180, 145, 205, 43.7, 99.9, 9.9, 3322.0],
    [0, 145, 205, 43.0, 80, 9.5, 3000.0],
    [180, 5, 5, 2.0, 15.0, 3.6, 10.0]
]

extreme_df = pd.DataFrame(extreme_tests, columns=features)

extreme_predictions = model.predict(extreme_df)

extreme_df["Predicted Crop"] = extreme_predictions

print(extreme_df.to_string(index=False))
```
